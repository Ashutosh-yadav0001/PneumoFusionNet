# MIMIC-CXR Radiology Reports Extractor

This notebook copies the free-text radiology reports corresponding to the 1,000 studies in `1mimic_subset_1000.csv` from your full downloaded reports folder into a new `subset_reports` folder.

```python
import os
import shutil
import pandas as pd
from tqdm.notebook import tqdm
```

## 1. Define Paths
We load the 1,000 patient subset CSV and specify the source reports directory and the destination folder.

```python
subset_path = '1mimic_subset_1000.csv'
reports_source_dir = os.path.join('restricted_Dowloaded_dataset', 'mimic-cxr-reports', 'files')
reports_dest_dir = os.path.join('restricted_Dowloaded_dataset', 'subset_reports')

df_subset = pd.read_csv(subset_path)
print(f"Loaded subset cohort with {len(df_subset)} patient studies.")
print(f"Source reports directory: {reports_source_dir}")
print(f"Destination reports directory: {reports_dest_dir}")
```

## 2. Extract and Copy Reports
We loop through your 1,000 patient subset, locate each study's text report, and copy it to `subset_reports` while preserving the patient folder hierarchy.

```python
success_count = 0
missing_count = 0

# Copy matching report files
for _, row in tqdm(df_subset.iterrows(), total=len(df_subset), desc="Copying Reports"):
    sub_id = str(int(row['subject_id']))
    std_id = str(int(row['study_id']))
    prefix = sub_id[:2]
    
    # Source report path: mimic-cxr-reports/files/pXX/pXXXXXXXX/sXXXXXXXX.txt
    relative_path = os.path.join(f"p{prefix}", f"p{sub_id}", f"s{std_id}.txt")
    source_file = os.path.join(reports_source_dir, relative_path)
    
    # Destination path: subset_reports/pXX/pXXXXXXXX/sXXXXXXXX.txt
    dest_file = os.path.join(reports_dest_dir, relative_path)
    
    if os.path.exists(source_file):
        # Recreate patient subfolders if necessary
        os.makedirs(os.path.dirname(dest_file), exist_ok=True)
        shutil.copy2(source_file, dest_file)
        success_count += 1
    else:
        print(f"Warning: Report not found for patient {sub_id}, study {std_id} at: {source_file}")
        missing_count += 1

print(f"\nReport copying completed!")
print(f"- Successfully copied: {success_count}")
print(f"- Missing reports: {missing_count}")
```
